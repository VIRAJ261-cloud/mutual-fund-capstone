```python
import sqlite3
import pandas as pd 
import numpy as np
from sqlalchemy import create_engine
```

## Top 5 funds by AUM

```python
engine = create_engine("sqlite:///../bluestock_mf.db")
conn = sqlite3.connect("../bluestock_mf.db")
cursor = conn.cursor()
query = """
SELECT
    df.scheme_name,
    fp.aum_crore
FROM fact_performance fp
JOIN dim_fund df
ON fp.fund_id = df.fund_id
ORDER BY fp.aum_crore DESC
LIMIT 5;
"""

pd.read_sql(query, engine)
```

## Average Nav per month

```python
query = """
SELECT
    dd.year,
    dd.month,
    ROUND(AVG(fn.nav),2) AS average_nav
FROM fact_nav fn
JOIN dim_date dd
ON fn.date_id = dd.date_id
GROUP BY dd.year, dd.month
ORDER BY dd.year, dd.month;
"""
pd.read_sql(query, engine)
```

## Total transaction by state

```python
query = """
SELECT
    state,
    COUNT(*) AS total_transactions
FROM fact_transactions
GROUP BY state
ORDER BY total_transactions DESC;
"""
pd.read_sql(query,engine)
```

## Total Transaction Amount by State

```python
query = """SELECT
    state,
    ROUND(SUM(amount_inr),2) AS total_amount
FROM fact_transactions
GROUP BY state
ORDER BY total_amount DESC;

"""
pd.read_sql(query,engine)
```

## Funds with Expense Ratio less than 1%

```python
query = """SELECT
    df.scheme_name,
    fp.expense_ratio_pct
FROM fact_performance fp
JOIN dim_fund df
ON fp.fund_id = df.fund_id
WHERE fp.expense_ratio_pct < 1
ORDER BY fp.expense_ratio_pct;
"""
pd.read_sql(query,engine)
```

## Top 10 Funds by Sharpe Ratio

```python
query = """
SELECT
    df.scheme_name,
    fp.sharpe_ratio
FROM fact_performance fp
JOIN dim_fund df
ON fp.fund_id = df.fund_id
ORDER BY fp.sharpe_ratio DESC
LIMIT 10;
"""
pd.read_sql(query,engine)
```

## Average 3-Year Return by Category

```python
query = """SELECT
    df.category,
    ROUND(AVG(fp.return_3yr_pct),2) AS avg_return_3yr
FROM fact_performance fp
JOIN dim_fund df
ON fp.fund_id = df.fund_id
GROUP BY df.category
ORDER BY avg_return_3yr DESC;"""
pd.read_sql(query,engine)
```

## Transactions by Payment Mode

```python
query = """SELECT
    payment_mode,
    COUNT(*) AS total_transactions,
    ROUND(SUM(amount_inr),2) AS total_amount
FROM fact_transactions
GROUP BY payment_mode
ORDER BY total_amount DESC;"""
pd.read_sql(query,engine)
```

## Number of Funds in each Risk Category

```python
query = """SELECT
    risk_category,
    COUNT(*) AS total_funds
FROM dim_fund
GROUP BY risk_category
ORDER BY total_funds DESC;
"""
pd.read_sql(query,engine)
```

## Average NAV by Fund House

```python
query = """SELECT
    df.fund_house,
    ROUND(AVG(fn.nav),2) AS average_nav
FROM fact_nav fn
JOIN dim_fund df
ON fn.fund_id = df.fund_id
GROUP BY df.fund_house
ORDER BY average_nav DESC;"""
pd.read_sql(query,engine)
```
