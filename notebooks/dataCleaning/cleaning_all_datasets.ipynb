# Creation of column that flags the anomaly

```python
df['exit_load_pct_anomaly']= df['exit_load_pct'].apply(lambda g:True if (g<lower_bound or g>upper_bound) else False)
```

```python
df.shape
```

```python
def flag_anomaly(columns):
    for column in columns:
        Q1 = df[column].describe()['25%']
        Q3 = df[column].describe()['75%']
        lower_bound = Q1 - (1.5*(Q3-Q1))
        upper_bound  = Q3 + (1.5*(Q3-Q1))
        df[f"{column}_anomaly"] = (
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        )
        if df[df[f"{column}_anomaly"]==True].empty:
            df.drop(f"{column}_anomaly", axis=1,inplace=True)
```

```python
flag_anomaly(df.describe().columns)
```
